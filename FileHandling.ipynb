```python
file=open(r"C:\Users\dhruv\OneDrive\Desktop\Folder\student.txt","r+")

data=file.read()

print("Old data : ",data)

file.seek(0) # pointer get position at start of the file 

file.write("HELLO")

file.seek(0)

x = file.read()
print("New Data:", x)

file.close()
```
